```python
import cv2
import subprocess
import numpy as np
 
# URL del stream
stream_url = "https://3-16-180-153.dash.rhombussystems.com:443/dash/api/outbound/EvOdvPdFTWiUiLsQC_TMKA/7GD8jLxZS0Sr7BesS94lqA/1739824479/live-v2/file.m3u8"
api_key = API_KEY.copy()
 
# Comando FFmpeg para acceder al stream con API Key
ffmpeg_cmd = [
    "ffmpeg",
    "-headers", f"Authorization: Bearer {api_key}",  # Agregar API Key
    "-i", stream_url,          # Entrada del stream
    "-f", "rawvideo",          # Salida en formato raw
    "-pix_fmt", "bgr24",       # Formato compatible con OpenCV
    "-vf", "fps=30",           # Ajustar FPS
    "-an", "-",                # Sin audio
]
 
# Ejecutar FFmpeg
ffmpeg_process = subprocess.Popen(ffmpeg_cmd, stdout=subprocess.PIPE, stderr=subprocess.DEVNULL, bufsize=10**8)
 
# Ajusta la resolución del video (según el stream)
frame_width = 1280  
frame_height = 720  
frame_size = frame_width * frame_height * 3  # 3 canales (BGR)
 
while True:
    # Leer los bytes del frame desde FFmpeg
    raw_frame = ffmpeg_process.stdout.read(frame_size)
    
    if len(raw_frame) != frame_size:
        print("Error: Frame incompleto")
        break
 
    # Convertir los bytes en un array de NumPy
    frame = np.frombuffer(raw_frame, np.uint8).reshape((frame_height, frame_width, 3))
 
    # Mostrar con OpenCV
    cv2.imshow("Stream con API Key", frame)
 
    # Salir con la tecla 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
 
# Cerrar procesos
ffmpeg_process.kill()
cv2.destroyAllWindows()
```
